```python
import networkx as nx
import matplotlib.pyplot as plt


G = nx.DiGraph()
G.add_edge("You", "Ana", weight=25, reason="pizza", currency="RON")     # pizza
G.add_edge("Ana", "Bogdan", weight=10, reason="coffee", duedate="tomorrow")  # coffee
G.add_edge("Bogdan", "You", weight=15)  # round 3
pos = nx.spring_layout(G)  # computes nodes' position once
nx.draw(G, pos, with_labels=True, node_color="green", node_size=1500, arrows=True)

edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.show()
```
